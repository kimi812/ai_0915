```python
# -----------------保留中文文字
import urllib.request
import re

# 從指定網址讀取 HTML 原始碼
x = urllib.request.urlopen('https://www.cycu.edu.tw/')
html = x.read().decode('utf-8')  # 以 UTF-8 解碼成字串

# 函式：移除所有 HTML 標籤
def remove_html_tags(text):
    clean = re.compile('<.*?>')   # 使用正規表達式匹配 HTML 標籤
    return re.sub(clean, '', text)  # 將 HTML 標籤替換為空字串

# 將 HTML 標籤移除
html = remove_html_tags(html)

# 移除所有空白字元（包含空格、換行、tab 等）
html = re.sub(r'\s', '', html)

# 建立正規表達式：匹配所有「不是中文」的字元
mych = re.compile(r'[^\u4e00-\u9fa5]')

# 將所有非中文字元移除，僅保留中文
html = re.sub(mych, '', html)

# 輸出結果
print(html)
```

```python
#------------------------統計多少個中文字-----------------

import urllib.request
import re

# 讀取網頁內容
x = urllib.request.urlopen('https://www.cycu.edu.tw/')
html = x.read().decode('utf-8')

# 函式：移除 HTML 標籤
def remove_html_tags(text):
    clean = re.compile('<.*?>')
    return re.sub(clean, '', text)

# 移除 HTML 標籤
html = remove_html_tags(html)

# 移除空白字元（空格、換行）
html = re.sub(r'\s', '', html)

# 保留「中文」，移除「所有非中文」字元
html = re.sub(r'[^\u4e00-\u9fa5]', '', html)

# 將每個中文字存到 list
mydata = []
for i in range(len(html)):
    mydata.append(html[i])

# 用 set 取得「不重複」的中文字集合
y = set(mydata)

print("中文種類：", y)
print("共有", len(y), "種不同的中文字")
```

```python
import urllib.request
import pandas as pd
import re
x = urllib.request.urlopen('https://www.cycu.edu.tw/')
html = x.read().decode('utf-8')
def remove_html_tags(text):
  clean = re.compile('<.*?>')
  return re.sub(clean, '', text)

html=remove_html_tags(html)
html = re.sub(r'\s', '', html)
mych=re.compile(r'[^\u4e00-\u9fa5]')
html=re.sub(mych, '',html)
mydata=[]
myvalue=[]
for i in range(len(html)):
  mydata.append(html[i])
mylist = list(dict.fromkeys(mydata))
for x in mylist:
  i = html.find(x)
  j=0
  while i != -1:
   j+=1
   i = html.find(x, i+1)
  myvalue.append(j)
df = pd.DataFrame(list(zip(mylist, myvalue)), columns =['單字', '次數'])
print(df)
```

```python
#-----------------前三名繪橫條圖
import urllib.request
import pandas as pd
import re
import matplotlib as mpl
import matplotlib.pyplot as plt
from google.colab import files

# ---- 指定中文字型 ------------------------------------------------------
mpl.rcParams["font.family"] = ["Noto Sans CJK JP"]
mpl.rcParams["axes.unicode_minus"] = False


x = urllib.request.urlopen('https://www.cycu.edu.tw/')
html = x.read( ).decode('utf-8')
def remove_html_tags(text):
  clean = re.compile('<.*?>')
  return re.sub(clean, '', text)

html=remove_html_tags(html)
html = re.sub(r'\s', '', html)
mych=re.compile(r'[^\u4e00-\u9fa5]')
html=re.sub(mych, '',html)
mydata=[]
myvalue=[]
for i in range(len(html)):
  mydata.append(html[i])
mylist = list(dict.fromkeys(mydata))
for x in mylist:
  i = html.find(x)
  j=0
  while i != -1:
   j+=1
   i = html.find(x, i+1)
  myvalue.append(j)
df = pd.DataFrame(list(zip(mylist, myvalue)), columns =['單字', '次數'])
dp=df.nlargest(6,'次數')
plt.barh(dp['單字'],dp['次數'],color="green")
plt.show()
```
